## 6. Payment Data Validation

Before cleaning the payment data, we will:
- Inspect its structure
- Check data types
- Check missing values
- Check duplicate payments
- Check payment methods and statuses
- Check whether every payment belongs to a valid order
- Check payment amount distribution

```python
payments = pd.read_csv("../data/raw/payments.csv")

print("Payments dataset loaded successfully!")
print("Shape:", payments.shape)

payments.head()
```

```python
payments.info()
```

```python
payments.describe(include="all")
```

```python
print("Missing values:")
print(payments.isnull().sum())
```

```python
print("Duplicate rows:", payments.duplicated().sum())
```

```python
print("Payment methods:")
print(payments["payment_method"].value_counts())

print("\nPayment statuses:")
print(payments["payment_status"].value_counts())
```

```python
invalid_payments = payments[
    ~payments["order_id"].isin(orders["order_id"])
]

print("Invalid payment records:", len(invalid_payments))
```

```python
print("Minimum payment:", payments["payment_amount"].min())
print("Maximum payment:", payments["payment_amount"].max())
print("Average payment:", payments["payment_amount"].mean())
print("Total payments:", payments["payment_amount"].sum())
```

## 6.1 Investigating Zero-Value Payments

A payment amount of zero may indicate a failed, cancelled, refunded,
or otherwise invalid transaction.

We will investigate the associated payment status before deciding
whether these records should be removed or retained.

```python
zero_payments = payments[payments["payment_amount"] == 0]

print("Number of zero-value payments:", len(zero_payments))

zero_payments.head(10)
```

```python
zero_payments["payment_status"].value_counts()
```

```python
zero_payments[["payment_status", "payment_method"]].value_counts()
```

```python
payments.groupby("payment_status")["payment_amount"].agg(
    ["count", "min", "max", "mean", "sum"]
)
```

## 6.2 Payment and Order ID Validation

Payment IDs and Order IDs act as identifiers and relationships
between our datasets.

We will check:
- Duplicate payment IDs
- Missing payment IDs
- Missing order IDs
- Whether every payment belongs to a valid order
- Whether one order has multiple payment records

```python
print("Duplicate payment IDs:", payments["payment_id"].duplicated().sum())
```

```python
print("Missing payment IDs:", payments["payment_id"].isnull().sum())
print("Missing order IDs:", payments["order_id"].isnull().sum())
```

```python
invalid_order_payments = payments[
    ~payments["order_id"].isin(orders["order_id"])
]

print("Payments with invalid order IDs:", len(invalid_order_payments))
```

```python
payments_per_order = payments.groupby("order_id").size()

print("Orders with multiple payment records:",
      (payments_per_order > 1).sum())

print("\nMaximum payments for a single order:",
      payments_per_order.max())
```

# 7. Overall Data Quality Assessment

Before beginning data cleaning, we will perform a consolidated
quality assessment across all ShopSphere datasets.

The assessment will examine:
- Dataset size
- Missing values
- Duplicate records
- Unique identifiers
- Referential integrity
- Numerical ranges
- Categorical values
- Potential business-rule violations

```python
datasets = {
    "Customers": customers,
    "Products": products,
    "Orders": orders,
    "Order Items": order_items,
    "Payments": payments
}

overview = []

for name, df in datasets.items():
    overview.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

quality_overview = pd.DataFrame(overview)

quality_overview
```

```python
for name, df in datasets.items():
    print(f"\n{'='*50}")
    print(name)
    print(f"{'='*50}")
    
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    
    if len(missing) == 0:
        print("No missing values found.")
    else:
        print(missing)
```

```python
for name, df in datasets.items():
    print(f"{name}: {df.duplicated().sum()} duplicate rows")
```

```python
id_columns = {
    "Customers": "customer_id",
    "Products": "product_id",
    "Orders": "order_id",
    "Order Items": "order_item_id",
    "Payments": "payment_id"
}

for name, id_col in id_columns.items():
    df = datasets[name]
    
    print(
        f"{name} | "
        f"Rows: {len(df)} | "
        f"Unique {id_col}: {df[id_col].nunique()} | "
        f"Missing: {df[id_col].isnull().sum()}"
    )
```

```python
print(
    "Orders with invalid customer IDs:",
    (~orders["customer_id"].isin(customers["customer_id"])).sum()
)

print(
    "Order items with invalid order IDs:",
    (~order_items["order_id"].isin(orders["order_id"])).sum()
)

print(
    "Order items with invalid product IDs:",
    (~order_items["product_id"].isin(products["product_id"])).sum()
)

print(
    "Payments with invalid order IDs:",
    (~payments["order_id"].isin(orders["order_id"])).sum()
)
```

```python
print("=" * 60)
print("SHOPSPHERE DATA QUALITY SUMMARY")
print("=" * 60)

for name, df in datasets.items():
    print(f"\n{name}")
    print(f"Rows              : {len(df):,}")
    print(f"Columns           : {df.shape[1]}")
    print(f"Missing values    : {df.isnull().sum().sum():,}")
    print(f"Duplicate rows    : {df.duplicated().sum():,}")
```
